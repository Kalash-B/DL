# Assignment 1: Study of Deep Learning Packages
**Aim:** Study of TensorFlow, Keras, Theano and PyTorch.

| Package | Developed by | Key idea |
|---|---|---|
| TensorFlow | Google | Tensors + computation graphs, production deployment (TF Serving, TF Lite) |
| Keras | F. Chollet (now part of TF) | High-level API: `Sequential`, `Dense`, `compile`, `fit` |
| Theano | Univ. of Montreal | Symbolic math compiler (discontinued, base for early Keras) |
| PyTorch | Meta (Facebook) | Dynamic computation graph, Pythonic, popular in research |

```python
import tensorflow as tf
from tensorflow import keras
import numpy as np
print("TensorFlow version:", tf.__version__)
print("Keras version:", keras.__version__)
```

### TensorFlow: tensors and basic operations

```python
a = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
b = tf.constant([[5, 6], [7, 8]], dtype=tf.float32)
print("a + b =\n", tf.add(a, b).numpy())
print("a * b (matmul) =\n", tf.matmul(a, b).numpy())

# Automatic differentiation: d/dx (x^2) at x = 3 -> 6
x = tf.Variable(3.0)
with tf.GradientTape() as tape:
    y = x ** 2
print("dy/dx =", tape.gradient(y, x).numpy())
```

### Keras: a small model (learns y = 2x + 1)

```python
X = np.arange(-10, 10, 0.5, dtype=np.float32)
Y = 2 * X + 1

model = keras.Sequential([keras.Input(shape=(1,)), keras.layers.Dense(1)])
model.compile(optimizer="sgd", loss="mse")
model.fit(X, Y, epochs=100, verbose=0)
print("Prediction for x=5:", model.predict(np.array([[5.0]]), verbose=0)[0][0])
model.summary()
```

### PyTorch (runs only if torch is installed)

```python
try:
    import torch
    print("PyTorch version:", torch.__version__)
    t = torch.tensor([[1., 2.], [3., 4.]], requires_grad=True)
    out = (t ** 2).sum()
    out.backward()
    print("Gradient:\n", t.grad)
except ImportError:
    print("PyTorch not installed (pip install torch)")
```
